```python
# Cell 1: Setup & Imports
%load_ext autoreload
%autoreload 2

from module_gates import (
    apply_x,
    apply_z,
    apply_h,
    apply_cnot,
    apply_toffoli,
    print_statevector,
)
```

```python
# Cell 2: Single-Qubit Gates (X, Z, H)
print("=================== 1-QUBIT GATES ===================")

# Pauli-X (Bit-Flip)
print_statevector(apply_x("0"), label="Pauli-X on |0>")
print_statevector(apply_x("1"), label="Pauli-X on |1>")

# Pauli-Z (Phase-Flip)
print_statevector(apply_z("0"), label="Pauli-Z on |0>")
print_statevector(apply_z("1"), label="Pauli-Z on |1>")

# Hadamard (Superposition)
print_statevector(apply_h("0"), label="Hadamard on |0> (|+> state)")
print_statevector(apply_h("1"), label="Hadamard on |1> (|-> state)")
```

```python
# Cell 3: Two-Qubit Controlled-NOT (CNOT)
print("=================== 2-QUBIT CNOT ===================")
# Qiskit qubit ordering is |q1 q0> (little-endian)
for state in ["00", "01", "10", "11"]:
    sv_res = apply_cnot(initial_state=state, control=0, target=1)
    print_statevector(sv_res, label=f"CNOT (ctrl=q0, tgt=q1) on |{state}>")
```

```python
# Cell 4: Three-Qubit Toffoli (CCX)
print("=================== 3-QUBIT TOFFOLI ===================")
# Toffoli flips target (q2) when both controls (q0, q1) are |1>
for state in ["000", "011", "110", "111"]:
    sv_res = apply_toffoli(initial_state=state, c1=0, c2=1, target=2)
    print_statevector(sv_res, label=f"Toffoli (ctrl=q0,q1, tgt=q2) on |{state}>")
```
